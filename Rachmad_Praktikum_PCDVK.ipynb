Step 1

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
```

Step 2

```python
#membuat list untuk menyimpan url
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg" , "https://picsum.photos/800/600" ,"https://picsum.photos/800/600?random=2" ]
for url in urls:
    print("Mencoba:", url)

    image = io.imread(url)

    image = cv.resize(image, (0, 0), fx=0.5, fy=0.5)

    image_2 = cv.cvtColor(image, cv.COLOR_RGB2BGR)

    final_frame = cv.hconcat((image, image_2))

    cv2_imshow(final_frame)
    print()
```

Langkah 3

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]

print("Tinggi:", tinggi)
print("Lebar:", lebar)

cv2_imshow(image_2)
```

Langkah 4

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range(lebar):
    image_3[int(tinggi / 2), y] = [255, 255, 255]


final_frame = cv.hconcat((image_2, image_3))

cv2_imshow(final_frame)
```
